```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

print("Libraries imported successfully!")
```

```python
df = pd.read_csv('/content/Delivery_Logistics_55_45.csv')

print("Dataset loaded successfully!")
```

```python
df['expected_time_hours'] = pd.to_numeric(
    df['expected_time_hours'].astype(str).str.extract(r'\.(\d+)$')[0],
    errors='coerce'
)

print(df['expected_time_hours'].head())
```

```python
df['is_late'] = df['delayed'].map({
    'no': 0,
    'yes': 1
})

print(df['is_late'].value_counts())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='delayed',
    hue='delayed',

    legend=False
)

plt.xticks([0, 1], ['On-Time', 'Delayed'])

plt.title("On-Time vs Delayed Deliveries")
plt.xlabel("Delivery Status")
plt.ylabel("Number of Deliveries")

plt.show()
```

```python
X = df.drop(columns=[
    'delivery_id',
    'delayed',
    'is_late',
    'delivery_status',
    'delivery_time_hours',
    'delivery_rating'
])

y = df['is_late']

print("Independent variables:", X.shape)
print("Dependent variable:", y.shape)
```

```python
X = pd.get_dummies(X, drop_first=True)

print(X.head())
print("\nTotal features:", X.shape[1])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

print("RANDOM FOREST RESULTS")

print("Accuracy:", accuracy_score(y_test, rf_pred))
print("Precision:", precision_score(y_test, rf_pred))
print("Recall:", recall_score(y_test, rf_pred))
print("F1 Score:", f1_score(y_test, rf_pred))

print("\nClassification Report:")

print(classification_report(y_test, rf_pred,target_names=['On-Time', 'Delayed']
))
```

```python
lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_model.fit(X_train_scaled, y_train)

lr_pred = lr_model.predict(X_test_scaled)

print("LOGISTIC REGRESSION RESULTS")

print("Accuracy:", accuracy_score(y_test, lr_pred))
print("Precision:", precision_score(y_test, lr_pred))
print("Recall:", recall_score(y_test, lr_pred))
print("F1 Score:", f1_score(y_test, lr_pred))

print("\nClassification Report:")
print(classification_report(y_test, lr_pred))
```

```python
linear_model = LinearRegression()

linear_model.fit(X_train_scaled, y_train)

linear_predictions = linear_model.predict(X_test_scaled)

linear_pred = (linear_predictions >= 0.5).astype(int)

print("LINEAR REGRESSION RESULTS")

print("Accuracy:", accuracy_score(y_test, linear_pred))
print("Precision:", precision_score(y_test, linear_pred))
print("Recall:", recall_score(y_test, linear_pred))
print("F1 Score:", f1_score(y_test, linear_pred))

print("\nClassification Report:")
print(classification_report(y_test, linear_pred))
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train_scaled, y_train)

knn_pred = knn_model.predict(X_test_scaled)

print("KNN RESULTS")

print("Accuracy:", accuracy_score(y_test, knn_pred))
print("Precision:", precision_score(y_test, knn_pred))
print("Recall:", recall_score(y_test, knn_pred))
print("F1 Score:", f1_score(y_test, knn_pred))

print("\nClassification Report:")
print(classification_report(y_test, knn_pred))
```

```python
svm_model = SVC(
    kernel='rbf',
    random_state=42
)

svm_model.fit(X_train_scaled, y_train)

svm_pred = svm_model.predict(X_test_scaled)

print("SVM RESULTS")

print("Accuracy:", accuracy_score(y_test, svm_pred))
print("Precision:", precision_score(y_test, svm_pred))
print("Recall:", recall_score(y_test, svm_pred))
print("F1 Score:", f1_score(y_test, svm_pred))

print("\nClassification Report:")
print(classification_report(y_test, svm_pred))
```

```python
comparison = pd.DataFrame({
    'Model': [
        'Random Forest',
        'Logistic Regression',
        'Linear Regression',
        'KNN',
        'SVM'
    ],

    'Accuracy': [
        accuracy_score(y_test, rf_pred),
        accuracy_score(y_test, lr_pred),
        accuracy_score(y_test, linear_pred),
        accuracy_score(y_test, knn_pred),
        accuracy_score(y_test, svm_pred)
    ],

    'Precision': [
        precision_score(y_test, rf_pred),
        precision_score(y_test, lr_pred),
        precision_score(y_test, linear_pred),
        precision_score(y_test, knn_pred),
        precision_score(y_test, svm_pred)
    ],

    'Recall': [
        recall_score(y_test, rf_pred),
        recall_score(y_test, lr_pred),
        recall_score(y_test, linear_pred),
        recall_score(y_test, knn_pred),
        recall_score(y_test, svm_pred)
    ],

    'F1 Score': [
        f1_score(y_test, rf_pred),
        f1_score(y_test, lr_pred),
        f1_score(y_test, linear_pred),
        f1_score(y_test, knn_pred),
        f1_score(y_test, svm_pred)
    ]
})

comparison
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=comparison,
    x='Model',
    y='Accuracy',
    hue='Model',
    palette='viridis',
    legend=False
)

plt.title("Accuracy Comparison of Five Models")
plt.xlabel("Machine Learning Models")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.xticks(rotation=20)

plt.show()
```
